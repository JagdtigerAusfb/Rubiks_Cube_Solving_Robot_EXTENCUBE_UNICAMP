```python
!pip install kociemba

import kociemba

state = (
    "LFURUBUUB"  # U
    "DDFBRUFDD"  # R
    "LBRRFLFFL"  # F
    "LRDDDLBLR"  # D
    "DBBULDUFU"  # L
    "RUBLBRFFR"  # B
)

# Embaralhamento exemplo: U D F2 R2 F' R2 L B' U L2 U2 L2 B' U2 B D2 F' L2 F2 D2 Rw2 Uw2

def count_moves(solution_str):
    if not solution_str.strip():
        return 0
    moves = solution_str.strip().split()
    return len(moves)

solution = kociemba.solve(state)
print("Solução:")
print(solution)
print("Número de movimentos:", count_moves(solution))
```
